````python
products = ['t-shirt', 'mug','hat','book','keychain']



def initialize_inventory(products):
            inventory = {product: int(input(f"Enter the quantity of {product}s available: ")) for product in products}
            return inventory



def get_customer_orders (): 
     customer_orders = {input('Enter the name of a product that you want to order: ') for x in range (int(input('How many products do you want to order? '))) }
    
     return customer_orders

#created dictionary

def create_order_dict (customer_orders):
     order_dict = {product_ordered: int(input(f'Enter the price of {product_ordered}: ')) for product_ordered in customer_orders}
     return order_dict

def sum_price_products (order_dict):
     total_cost = sum(order_dict.values())
     return total_cost
     

def update_inventory (customer_orders, inventory):
    #step one to delete the number of products
    for product in customer_orders:
       inventory [product] -=1
    #step 2 rebuild inventory using comprehension -- seems easier to just delete the product in the inventory with an if but I have to use comprehension and rebuild the inventory
    inventory = {product:quantity for product , quantity in inventory.items() if quantity > 0  }
    return inventory



def calculate_order_statistics (customer_orders,products):
    total_products_ordered = len(customer_orders)
    percentage_ordered = (total_products_ordered / len(products)) * 100
    order_statistics = (total_products_ordered,   percentage_ordered)
    return order_statistics 




def print_order_statistics (order_statistics):
    print (f'The total products ordered are: {order_statistics[0]} which is {order_statistics[1]}%')
    return 


def print_updated_inventory (inventory):
    print('The inventory now has:')
    for product in inventory :
        print ( f'{inventory [product]} {product}')





````

````python
inventory = initialize_inventory (products)
customer_orders = get_customer_orders ()
order_dict = create_order_dict (customer_orders)
total_cost = sum_price_products (order_dict)
inventory = update_inventory(customer_orders, inventory)
order_statistics = calculate_order_statistics (customer_orders, products)
print()
print_order_statistics (order_statistics)
print()
print_updated_inventory (inventory)
print (f' Total price: {total_cost}')
````
